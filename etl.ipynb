```python
#menginport library pandas untuk melakukan manipulasi dan analisis data
#membuat data frame dari dictionary
import pandas as pd

#sumber data yang mencakup table id barang, nama barang, jumlah terjual, dan harga per unit

data = {
    'idbarang': ['1', '2', '3', '4', '5'],
    'namabarang': ['barang a', 'barang b', 'barang c', 'barang d', 'barang e'],
    'jumlah_terjual': [10, 15, 7, 20, 5],
    'harga_per_unit': [1000, 1500, 700, 2000, 500]
}

#membuat data frame dari dictionary data, lalu mengkonversi data menjadi table yang terstruktur
df = pd.DataFrame(data)

#membuat kolom baru yang berisi total penjualan
#dengan mengalikan jumlah terjual dengan harga per unit
df['total_penjualan'] = df['jumlah_terjual'] * df['harga_per_unit']

#membuat data frame baru yang hanya berisi kolom id barang, nama barang, dan total penjualan
df_hasil = df[['idbarang', 'namabarang', 'total_penjualan']]

#menampilkan data frame hasil
print(df_hasil)
```

```python
import pandas as pd
import numpy as np
# Data sumber (Extract)
data = {
    'ID_Barang': [1, 2, 3, 4, 5],
    'Nama_Barang': ['Barang A', 'Barang B', 'Barang C', 'Barang D', 'BarangE'],
    'Jumlah_Terjual': [10, np.nan, 7, 20, 5],
    'Harga_Per_Unit': [1000, 1500, np.nan, 2000, 500]
}

df = pd.DataFrame(data)
# Menampilkan dataset asli
print("Dataset Asli:")
print(df)
# Transformasi data
# Mengisi nilai kosong dengan nilai rata-rata
df['Jumlah_Terjual'].fillna(df['Jumlah_Terjual'].mean(), inplace=True)
df['Harga_Per_Unit'].fillna(df['Harga_Per_Unit'].mean(), inplace=True)
# Menghitung total penjualan
df['Total_Penjualan'] = df['Jumlah_Terjual'] * df['Harga_Per_Unit']
# Memuat data ke DataFrame baru (Load)
df_hasil = df[['ID_Barang', 'Nama_Barang', 'Total_Penjualan']]
# Menampilkan dataset hasil
print("\nDataset Hasil:")
print(df_hasil)
```
